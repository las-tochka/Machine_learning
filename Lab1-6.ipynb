4. Как найти самое часто встречаемое значение между двумя массивами? 

````python
# бужем считать, что ищем число, что есть в обоих массивах
# и встречается наиболее часто
arr1 = np.array([1, 2, 2, 2, 2, 3, 7])
arr2 = np.array([ 3, 4, 5])
common_elements = np.intersect1d(arr1, arr2)
combined = np.concatenate((arr1, arr2))
values, counts = np.unique(combined, return_counts=True)
mask = np.isin(values, common_elements)
filt_values = values[mask]
filt_counts = counts[mask]
most_frequent_common = filt_values[np.argmax(filt_counts)]
print(most_frequent_common)
````

8. Создайте случайную матрицу 10х2 которая представляет собой декартовые координаты и преобразуйте их в полярные

````python
cartesian = np.random.uniform(-10, 10, size=(10, 2))

x = cartesian[:, 0]
y = cartesian[:, 1]

r = np.sqrt(x**2 + y**2)
phi = np.arctan2(y, x)
polar = np.column_stack((r, phi))

print("Декартовы координаты (X, Y):\n", cartesian)
print("\nПолярные координаты (r, phi):\n", polar)
````

1. Посчитайте среднее кол-во часов уделяемое на сон студентами

````python
mean_sleep = df['Sleep_Hours'].mean()
print(f"Среднее количество часов на сон: {mean_sleep:.2f}")
````

2. Сколько процентов студентов посещают внекласные активности?

````python
count_yes = (df['Extracurricular_Activities'] == 'Yes').sum()

total_students = len(df)
percentage = (count_yes / total_students) * 100
print(f"Процент студентов с активностями: {percentage:.2f}%")
````

3. Правда ли, что в среднем девушки сдают экзамены лучше, чем мужчины

````python
mean_scores = df.groupby('Gender')['Exam_Score'].mean()
# print(mean_scores)
if mean_scores.get('Female', 0) > mean_scores.get('Male', 0):
    print("Ответ: Да, в среднем девушки сдают экзамены лучше.")
else:
    print("Ответ: Нет, в среднем мужчины сдают экзамены не хуже (или лучше).")
````

4. Выведите 25 и 75 перцентили времени сна студентов

````python
q25 = df['Sleep_Hours'].quantile(0.25)
q75 = df['Sleep_Hours'].quantile(0.75)
print(f"25-й перцентиль (Q1): {q25}")
print(f"75-й перцентиль (Q3): {q75}")
````

5. Правда ли, что те, кто спит больше(75 перцентиль и выше) сдает экзамен лучше, чем студенты с недосыпом(25 перцентиль и меньше)

````python
q75 = df['Sleep_Hours'].quantile(0.75)
q75_mean = df[df['Sleep_Hours'] >= q75]['Exam_Score'].mean()
q25 = df['Sleep_Hours'].quantile(0.25)
q25_mean = df[df['Sleep_Hours'] <= q25]['Exam_Score'].mean()
if q25_mean < q75_mean:
    print("Ответ: Да, в среднем люди с недосыпом сдают экзамены хуже.")
else:
    print("Ответ: Нет, в среднем высыпающиеся люди не сдают экзамены лучше")
````

6. Влияет ли физическая активность на результаты экзамена?

````python
mean_scores = df.groupby('Physical_Activity')['Exam_Score'].mean()
print(mean_scores)
best_pair = max(mean_scores.items(), key=lambda x: x[1])
print(f"Лучшая активность: {best_pair[0]}, балл: {best_pair[1]:.2f}")
````

7. Правда ли, что слуденты, чьи родители имеют высшее образование(College || Postgraduate), сдают экзамены лучше, чем остальные?

````python
mean_scores = df.groupby('Parental_Education_Level')['Exam_Score'].mean()
sorted_mean_scores = sorted(mean_scores.items(), key=lambda x: x[1])
# print(sorted_mean_scores, '\n',  mean_scores)
checked_val = ['Postgraduate', 'College']
if sorted_mean_scores[-1][0] in checked_val and sorted_mean_scores[-2][0] in checked_val:
    print("Да, образование родителей влияет на оценки")
else:
    print("Нет, образование родителей не влияет на оценки")
````

8. Как влияет участие родителей на результаты экзаменов

````python
mean_scores = df.groupby('Parental_Involvement')['Exam_Score'].mean()
sorted_mean_scores = sorted(mean_scores.items(), key=lambda x: x[1])
for level, score in sorted_mean_scores:
    print(f'Люди с уровнем {level} учестия родителей в среднеи получили такую {score} оценку')
````

9. Правда ли, что в семьях с доходом больше среднего дети сдают экзамены лучше?

````python
mean_scores = df.groupby('Family_Income')['Exam_Score'].mean()
high_score = mean_scores.get('High', 0)
other_scores = mean_scores.drop('High', errors='ignore').mean()
if high_score > other_scores:
    print("Да, доход решает")
else:
    print("Нет, упорство важнее")
````

10. Посчитайте долю студентов сдавших экзамен на более чем 70 баллов в семьях с низким доходом и отсутствием доступа к интернету

````python
target_students = df[(df['Family_Income'] == 'Low') & (df['Internet_Access'] == 'No')]
passed = (target_students['Exam_Score'] > 70).sum()
total = len(target_students)
share = (passed / total) * 100
print(f"Всего студентов Low + No Internet: {total}")
print(f"Из них сдали > 70: {passed}")
print(f"Доля: {share:.2f}%")
````

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

1. Правда ли, что количество часов затрачиваемых на сон распределено примерно нормально?

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid")
df = pd.read_csv('student_perfomance.csv')

# 1. Распределение часов сна
sns.histplot(data=df, x='Sleep_Hours', color='skyblue', discrete=True, stat='count', alpha=0.6)
counts = df['Sleep_Hours'].value_counts().sort_index()
plt.plot(counts.index, counts.values, color='red', marker='o', linestyle='-', linewidth=2, label='Полигон частот')

plt.title('Распределение часов сна (Полигон частот)', fontsize=14)
plt.xlabel('Часы сна', fontsize=12)
plt.ylabel('Количество студентов', fontsize=12)
plt.legend()
plt.show()
````

2. Постройте гистограмму распределения количества часов затрачиваемых на обучение среди мальчиков и девочек

````python
sns.histplot(data=df, x='Hours_Studied', hue='Gender', 
             multiple='dodge', bins=15, kde=True, palette='Set2')

plt.title('Распределение часов обучения среди мальчиков и девочек', fontsize=14)
plt.xlabel('Количество часов обучения в неделю', fontsize=12)
plt.ylabel('Количество студентов', fontsize=12)
plt.grid(True, alpha=0.3)

plt.show()
````

3. Как зависит результат конечного экзамена от результата первого экзамена

````python
sns.regplot(data=df, x='Previous_Scores', y='Exam_Score', 
            scatter_kws={'color': 'blue'}, 
            line_kws={'color': 'red'})

plt.title('Зависимость конечного балла от первого экзамена', fontsize=14)
plt.xlabel('Балл за первый экзамен (Previous_Scores)', fontsize=12)
plt.ylabel('Балл за конечный экзамен (Exam_Score)', fontsize=12)
plt.grid(True, alpha=0.3)

plt.show()
````

4. Постройте box plot отображающий распределение количества сна детей сдавших экзамены на оценку ниже и выше 85 баллов

````python
df['Score_Category'] = df['Exam_Score'].apply(lambda x: 'Выше 85' if x > 85 else 'Ниже 85')
plt.figure(figsize=(8, 6))
sns.boxplot(data=df,
    x='Score_Category',
    y='Sleep_Hours',
    hue='Score_Category',
    palette='Set2', 
    legend=False   )
plt.title('Распределение часов сна в зависимости от балла за экзамен', fontsize=14)
plt.xlabel('Результат экзамена', fontsize=12)
plt.ylabel('Количество часов сна', fontsize=12)
plt.grid(True, alpha=0.3)

plt.show()
````

5. Правда ли, что количество посещаемых занятий с репетитором в месяц напрямую влияет на балл в конечном экзамене?

````python
sns.regplot(data=df, x='Tutoring_Sessions', y='Exam_Score', 
            scatter_kws={'color': 'blue'},
            line_kws={'color': 'red'})

plt.title('Зависимость балла экзамена от занятий у репетитора', fontsize=14)
plt.xlabel('Количество посещенных занятий', fontsize=12)
plt.ylabel('Балл за экзамен', fontsize=12)
plt.grid(True, alpha=0.3)

plt.show()
````

6. Как физическая активность влияет на успеваемость?

````python
mean_scores = df.groupby('Physical_Activity')['Exam_Score'].mean().reset_index()

sns.lineplot(data=mean_scores, x='Physical_Activity', y='Exam_Score', 
             marker='o', color='blue', linewidth=2, markersize=8)

plt.title('Средний балл за экзамен в зависимости от физической активности', fontsize=14)
plt.xlabel('Уровень физической активности', fontsize=12)
plt.ylabel('Средний балл за экзамен', fontsize=12)
plt.show()
````

7. Постройте HeatMap где ось Х - процент посещаемость, Y - количество занятий с репетитором, а Z - балл за итоговый экзамен*

\* Задание на доп. балл

````python
df['Attendance_Bin'] = pd.cut(df['Attendance'], 
                              bins=[0, 60, 70, 80, 90, 100], 
                              labels=['0-60%', '60-70%', '70-80%', '80-90%', '90-100%'])

heatmap_data = df.pivot_table(index='Tutoring_Sessions', 
                              columns='Attendance_Bin', 
                              values='Exam_Score', 
                              aggfunc='mean')

plt.figure(figsize=(12, 8))

sns.heatmap(data=heatmap_data, cmap='YlGnBu', annot=True, fmt='.1f',
            linewidths=.5, cbar_kws={'label': 'Средний балл'})

plt.title('Средний балл за экзамен в зависимости от посещаемости и занятий с репетитором', fontsize=14)
plt.xlabel('Процент посещаемости', fontsize=12)
plt.ylabel('Количество занятий с репетитором', fontsize=12)

plt.show()
````
